```python
import pandas as pd
import plotly.graph_objects as go
%pip install nbformat
```

```python
bh = pd.read_csv("../output/AAPL_buyhold.csv")
mac = pd.read_csv("../output/AAPL_crossover_50_200.csv")
```

```python
cumMaxHold = bh["Equity"].cummax()
cumMaxCross = mac["Equity"].cummax()

ddHold  = (bh["Equity"]  - cumMaxHold)  / cumMaxHold
ddCross = (mac["Equity"] - cumMaxCross) / cumMaxCross
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=bh["Date"],  y=ddHold * 100,  name="Buy & Hold"))
fig.add_trace(go.Scatter(x=mac["Date"], y=ddCross * 100, name="MA Crossover"))
fig.update_layout(title="AAPL: Drawdown Over Time",
                  xaxis_title="Date", yaxis_title="Drawdown (%)")
fig.show()
```

While it isn't suprising that Buy and Hold over the years 2014-2024 way out performed Moving Average Crossover(50, 200), the graph shows us that crossover went up from 100k -> ~226k and then back down to ~123k in mid 2019 where it had its worst drawdown of 45%.

Crossover (50/200) on AAPL 2015-2024 returned ~407% vs buy-and-hold's ~928%, with a deeper max drawdown (~45.6% vs ~38.5%), peaked in Jan 2022 and never recovered while buy-and-hold set new highs through 2024.

This result is expected rather than a failure of the strategy: buy-and-hold is very hard to beat on a single asset in a sustained uptrend, since any time spent in cash means missed gains. Evaluating trend-following fairly requires testing across other regimes (declining, sideways, or high-volatility markets), which is the next phase of analysis.
